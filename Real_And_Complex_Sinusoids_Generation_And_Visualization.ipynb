```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Real Sinusoid Generation
#The formula of a real sinusoid is given by: A * cos(2 * pi * f * t + phi)
A=0.8 #Amplitude
f0=1000 #Frequency in Hz
fs=44100 #Sampling frequency in Hz
phi=np.pi/2 #Phase in radians
t=np.arange(-0.002, 0.002, 1/fs) #Time vector from -0.002 to 0.002 seconds with a sampling frequency of 44100 Hz
x_real=A * np.cos(2*np.pi*f0*t+phi) #Real sinusoid
```

```python
plt.plot(t, x_real)
plt.title('Real Sinusoid')
plt.axis([-0.002, 0.002, -0.8, 0.8])
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.show()
```

```python
#Complex Sinusoid Generation
#The formula of a complex sinusoid is given by: exp(j * (2 * pi * k * n / N ))
N = 500 #Number of samples
k = 5 #Frequency index or number of cycles in N samples
n = np.arange(-N/2, N/2) #Sample index from -N/2 to N/2 - 1
x_complex = np.exp(1j * (2 * np.pi * k * n / N)) #Complex sinusoid
```

```python
#Plotting the real part of the complex sinusoid
plot_real = plt.plot(n, np.real(x_complex))
plt.title('Complex Sinusoid (Real Part)')   
plt.axis([-N/2, N/2, -1, 1])
plt.xlabel('Sample Index (n)')
plt.ylabel('Amplitude')
plt.show(plot_real) #plt.grid()
```

```python
#Plotting the imaginary part of the complex sinusoid
plot_imaginary = plt.plot(n, np.imag(x_complex))
plt.title('Complex Sinusoid (Imaginary Part)')   
plt.axis([-N/2, N/2, -1, 1])
plt.xlabel('Sample Index (n)')
plt.ylabel('Amplitude')
plt.show(plot_imaginary) #plt.grid()
```
